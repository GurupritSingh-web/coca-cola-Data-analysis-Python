```python
revenue = df[df["Particular"].str.contains("NET OPERATING REVENUES", case=False)]
```

```python
#Revenue_Growth% = RevenueFY18 − RevenueFY09 / RevenueFY09 * 100
fy09 = revenue["FY '09"].iloc[0]
fy18 = revenue["FY '18"].iloc[0]

growth = ((fy18 - fy09) / fy09) * 100

print("FY'09 Revenue:", fy09)
print("FY'18 Revenue:", fy18)
print("Overall Revenue Growth:", round(growth, 2), "%")
```

```python
operating_income = df[df["Particular"] == "Operating Income"]

op_income_09 = operating_income["FY '09"].iloc[0]
op_income_18 = operating_income["FY '18"].iloc[0]

op_margin_09 = (op_income_09 / fy09) * 100
op_margin_18 = (op_income_18 / fy18) * 100

print("FY'09 Operating Margin:", round(op_margin_09, 2), "%")
print("FY'18 Operating Margin:", round(op_margin_18, 2), "%")
```

```python
net_income = df[
    df["Particular"] ==
    "NET INCOME ATTRIBUTABLE TO SHAREOWNERS OF THE COCA-COLA COMPANY"]

net09 = net_income["FY '09"].iloc[0]
net18 = net_income["FY '18"].iloc[0]

net_margin_09 = (net09 / fy09) * 100
net_margin_18 = (net18 / fy18) * 100

print("FY'09 Net Profit Margin:", round(net_margin_09, 2), "%")
print("FY'18 Net Profit Margin:", round(net_margin_18, 2), "%")
```

```python
COGS = df[df["Particular"].str.contains("Cost of goods sold", case=False)]
COGS
```

```python
cogs09 = COGS["FY '09"].iloc[0]
cogs18 = COGS["FY '18"].iloc[0]

cogs_pct_09 = (cogs09 / fy09) * 100
cogs_pct_18 = (cogs18 / fy18) * 100

print("FY'09 COGS % of Revenue:", round(cogs_pct_09, 2), "%")
print("FY'18 COGS % of Revenue:", round(cogs_pct_18, 2), "%")
```

```python
assets = df[df["Particular"] == "Total assets"]

assets09 = assets["FY '09"].iloc[0]
assets18 = assets["FY '18"].iloc[0]

asset_growth = ((assets18 - assets09) / assets09) * 100

print("FY'09 Total Assets:", assets09)
print("FY'18 Total Assets:", assets18)
print("Total Asset Growth:", round(asset_growth, 2), "%")
```

```python
years = df.columns[1:]

operating_margin = (
    operating_income[years].iloc[0] /
    revenue[years].iloc[0]
) * 100

print(operating_margin.round(2))
```

```python
change = operating_margin["FY '18"] - operating_margin["FY '09"]

print("Change in Operating Margin:",
      round(change, 2), "percentage points")
```

```python
revenue_values = revenue[years].iloc[0]

print("Highest Revenue:")
print(revenue_values.idxmax(), revenue_values.max())

print("\nLowest Revenue:")
print(revenue_values.idxmin(), revenue_values.min())
```

```python
operating_values = operating_income[years].iloc[0]

print("Highest Operating Income:")
print(operating_values.idxmax(), operating_values.max())

print("\nLowest Operating Income:")
print(operating_values.idxmin(), operating_values.min())
```

```python
plt.plot(years, revenue_values, marker="o")

plt.title("Coca-Cola Revenue Trend (FY'09–FY'18)")
plt.xlabel("Fiscal Year")
plt.ylabel("Revenue (Million USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.plot(
    years,
    operating_margin,
    marker="o"
)

plt.title("Coca-Cola Operating Profit Margin (FY'09–FY'18)")
plt.xlabel("Fiscal Year")
plt.ylabel("Operating Margin (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.plot(years, revenue_values, marker="o", label="Revenue")
plt.plot(years, operating_values, marker="o", label="Operating Income")

plt.title("Revenue vs Operating Income")
plt.xlabel("Fiscal Year")
plt.ylabel("Million USD")
plt.xticks(rotation=45)
plt.legend()

plt.show()
```
